## Calculating Momentum Signals

```python
# Calculate momentum signals (12-months trailing returns)
momentum_signals = monthly_returns.rolling(window=12).apply(lambda x: np.prod(1 + x) - 1).shift(1).dropna()

# Creating Portfolios (quintiles) based on momentum signals
momentum_portfolios = momentum_signals.rank(axis=1).apply(lambda x: pd.qcut(x, q=5, labels=False), axis=1)

# Compute the subsequent month's returns for each portfolio
portfolio_return = pd.DataFrame(index=monthly_returns.index, columns=range(5))

for month in range(len(momentum_signals)-1):
    for i in range(5):
        # Find countries in each portfolio
        portfolio_countries = momentum_signals.columns[momentum_portfolios.iloc[month] == 1]
        
        # Calculate the next month's returns for each portofolio
        portfolio_return.iloc[month + 1, i] = monthly_returns.iloc[month + 1] [portfolio_countries].mean()

# Calculate mean return, volatitlity and sharpe ratios for each portfolio
annual_portfolio_returns = portfolio_return.mean() * 12
annual_portfolio_volatility = portfolio_return.std() * np.sqrt(12)
portfolio_sharpe_ratios = annual_portfolio_returns / annual_portfolio_volatility


#Plotting cummulative returns
cummulative_returns = (1 + portfolio_return).cumprod()

plt.figure(figsize=(10,6))
for i in range(5):
    plt.plot(cummulative_returns.index, cummulative_returns[i], label=f'Portfolio{i+1}')
    plt.legend()
    plt.title('Cumulative Returns of Momentum Portfolios')
    plt.show()

# Print portfolio performance
porfolio_summary = pd.DataFrame({
    'Mean Returns': annual_portfolio_returns,
    'Volatility': annual_portfolio_volatility,
    'Sharpe Ratios': portfolio_sharpe_ratios
})
print(porfolio_summary)
```
